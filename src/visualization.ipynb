```python

import matplotlib.pyplot as plt
import numpy as np

# ==========================================
# Figure 1: Visualization of ATE
# 1. DATA
# ==========================================

variables = [
    "Customer Service Contact",
    "International Plan",
    "Voice Mail Plan"
]

y_pos = np.arange(len(variables))

ate = np.array([0.032498, 0.294161, -0.088740])

ci_lower = np.array([0.013898, 0.249988, -0.109390])
ci_upper = np.array([0.052194, 0.338329, -0.069750])

ate_texts = [
    "0.0325 [0.0139; 0.0522]",
    "0.2942 [0.2500; 0.3383]",
    "-0.0887 [-0.1094; -0.0698]"
]

# ==========================================
# 2. FIGURE
# ==========================================

fig, ax = plt.subplots(figsize=(12, 7), dpi=150)

main_color = "#19324D"

# Zero reference line
ax.axvline(
    x=0,
    color="#777777",
    linestyle="--",
    linewidth=1.3,
    zorder=1
)

# ==========================================
# 3. FOREST PLOT
# ==========================================

ax.errorbar(
    ate,
    y_pos,
    xerr=[
        ate - ci_lower,
        ci_upper - ate
    ],
    fmt="o",
    color=main_color,
    ecolor=main_color,
    elinewidth=2.5,
    capsize=6,
    markersize=9,
    markeredgecolor="white",
    markeredgewidth=1.2,
    zorder=3
)

# ==========================================
# 4. NUMBERS ABOVE CI BARS
# ==========================================

for i in range(len(variables)):
    ax.annotate(
        ate_texts[i],
        xy=(ate[i], y_pos[i]),
        xytext=(0, 16),
        textcoords="offset points",
        ha="center",
        va="bottom",
        fontsize=9,
        fontweight="bold",
        color=main_color,
        annotation_clip=False
    )

# ==========================================
# 5. AXES
# ==========================================

ax.set_yticks(y_pos)
ax.set_yticklabels(
    variables,
    fontsize=11,
    fontweight="bold"
)

ax.set_xlabel(
    "Average Treatment Effect (ATE)",
    fontsize=12,
    fontweight="bold",
    labelpad=12
)

ax.set_title(
    "Estimated Causal Effects on Customer Churn",
    fontsize=15,
    fontweight="bold",
    pad=22
)

ax.set_xlim(-0.19, 0.53)
ax.set_ylim(-0.7, 2.7)

ax.grid(
    axis="x",
    linestyle=":",
    linewidth=0.8,
    alpha=0.45
)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_color("#CCCCCC")
ax.spines["bottom"].set_color("#CCCCCC")

ax.tick_params(axis="x", labelsize=10)
ax.tick_params(axis="y", length=0)

# ==========================================
# 6. FOOTNOTE
# ==========================================

fig.text(
    0.12,
    0.015,
    "Note: Error bars represent 95% confidence intervals. "
    "The dashed vertical line indicates zero treatment effect.",
    fontsize=8.5,
    color="#555555"
)

plt.tight_layout(rect=[0, 0.05, 1, 1])

plt.savefig(
    "ate_results.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# Figure 2: Visualization of Placebo test

import matplotlib.pyplot as plt
import numpy as np

# ==========================================
# 1. DATA
# ==========================================

variables = [
    "Customer Service Contact",
    "International Plan",
    "Voice Mail Plan"
]

placebo_p = [0.90, 0.94, 0.86]
random_cause_p = [0.94, 0.80, 0.94]

x = np.arange(len(variables))
width = 0.32

# ==========================================
# 2. STYLE
# ==========================================

fig, ax = plt.subplots(figsize=(11, 6.5), dpi=150)

color_placebo = "#19324D"
color_random = "#B45A4A"

# ==========================================
# 3. BAR CHART
# ==========================================

bars1 = ax.bar(
    x - width / 2,
    placebo_p,
    width,
    label="Placebo Test",
    color=color_placebo,
    edgecolor="white",
    linewidth=1
)

bars2 = ax.bar(
    x + width / 2,
    random_cause_p,
    width,
    label="Random Common Cause",
    color=color_random,
    edgecolor="white",
    linewidth=1
)

# ==========================================
# 4. VALUE LABELS
# ==========================================

for bars in [bars1, bars2]:
    for bar in bars:
        height = bar.get_height()

        ax.annotate(
            f"{height:.2f}",
            xy=(bar.get_x() + bar.get_width() / 2, height),
            xytext=(0, 7),
            textcoords="offset points",
            ha="center",
            va="bottom",
            fontsize=10,
            fontweight="bold"
        )

# ==========================================
# 5. REFERENCE THRESHOLD
# ==========================================

ax.axhline(
    y=0.05,
    color="#777777",
    linestyle="--",
    linewidth=1.4,
    label="Reference threshold (0.05)"
)

# ==========================================
# 6. AXES AND LABELS
# ==========================================

ax.set_xticks(x)
ax.set_xticklabels(
    variables,
    fontsize=10.5,
    fontweight="bold"
)

ax.set_ylabel(
    "Refutation p-value",
    fontsize=11,
    fontweight="bold",
    labelpad=10
)

ax.set_title(
    "Causal Model Refutation Diagnostics",
    fontsize=15,
    fontweight="bold",
    pad=20
)

ax.set_ylim(0, 1.12)

ax.grid(
    axis="y",
    linestyle=":",
    alpha=0.4
)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.legend(
    loc="upper center",
    bbox_to_anchor=(0.5, -0.13),
    ncol=3,
    frameon=False,
    fontsize=9
)

ax.tick_params(axis="both", labelsize=10)

# ==========================================
# 7. FOOTNOTE
# ==========================================

fig.text(
    0.10,
    0.015,
    "Note: Higher p-values indicate that these particular "
    "refutation tests did not detect a statistically significant difference.",
    fontsize=8.5,
    color="#555555"
)

plt.tight_layout(rect=[0, 0.09, 1, 1])

plt.savefig(
    "diagnostics.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# Figure 3: Visualization of Sensitivity test


import matplotlib.pyplot as plt
import numpy as np

# ==========================================
# 1. DATA
# ==========================================

strengths = [0.01, 0.05, 0.10, 0.20, 0.30]

# International Plan
international = [
    0.336554,
    0.436999,
    0.536393,
    0.631271,
    0.665591
]

# Voice Mail Plan
voicemail = [
    -0.069183,
    -0.015117,
    0.053685,
    0.158061,
    0.226369
]

# Customer Service Contact
service_contact = [
    0.008720,
    -0.084511,
    -0.181006,
    -0.292780,
    -0.359610
]

# Original ATE
original_ate = {
    "International Plan": 0.294161,
    "Voice Mail Plan": -0.088740,
    "Customer Service Contact": 0.032498
}

# ==========================================
# 2. STYLE
# ==========================================

fig, ax = plt.subplots(figsize=(11, 7), dpi=150)

colors = {
    "International Plan": "#19324D",
    "Voice Mail Plan": "#B45A4A",
    "Customer Service Contact": "#477A68"
}

# ==========================================
# 3. SENSITIVITY LINES
# ==========================================

ax.plot(
    strengths,
    international,
    marker="o",
    markersize=7,
    linewidth=2.5,
    color=colors["International Plan"],
    label="International Plan"
)

ax.plot(
    strengths,
    voicemail,
    marker="s",
    markersize=7,
    linewidth=2.5,
    color=colors["Voice Mail Plan"],
    label="Voice Mail Plan"
)

ax.plot(
    strengths,
    service_contact,
    marker="D",
    markersize=6,
    linewidth=2.5,
    color=colors["Customer Service Contact"],
    label="Customer Service Contact"
)

# ==========================================
# 4. ORIGINAL ATE REFERENCE LINES
# ==========================================

for label, value in original_ate.items():
    ax.axhline(
        y=value,
        color=colors[label],
        linestyle=":",
        linewidth=1.2,
        alpha=0.65
    )

# ==========================================
# 5. ZERO REFERENCE LINE
# ==========================================

ax.axhline(
    y=0,
    color="#555555",
    linestyle="--",
    linewidth=1.4,
    label="Zero effect"
)

# ==========================================
# 6. VALUE LABELS
# ==========================================

for values, color, offset in [
    (international, colors["International Plan"], 9),
    (voicemail, colors["Voice Mail Plan"], 9),
    (service_contact, colors["Customer Service Contact"], -13)
]:

    for x_value, y_value in zip(strengths, values):
        ax.annotate(
            f"{y_value:.3f}",
            xy=(x_value, y_value),
            xytext=(0, offset),
            textcoords="offset points",
            ha="center",
            va="bottom" if offset > 0 else "top",
            fontsize=8,
            color=color,
            fontweight="bold"
        )

# ==========================================
# 7. AXES AND LABELS
# ==========================================

ax.set_xlabel(
    "Simulated Unobserved Confounding Strength",
    fontsize=11,
    fontweight="bold",
    labelpad=10
)

ax.set_ylabel(
    "Estimated ATE",
    fontsize=11,
    fontweight="bold",
    labelpad=10
)

ax.set_title(
    "Sensitivity Analysis for Unobserved Confounding",
    fontsize=15,
    fontweight="bold",
    pad=20
)

ax.set_xticks(strengths)
ax.set_xticklabels(
    [f"{s:.2f}" for s in strengths]
)

ax.set_xlim(0, 0.32)
ax.set_ylim(-0.45, 0.78)

ax.grid(
    linestyle=":",
    alpha=0.35
)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.legend(
    loc="upper left",
    frameon=False,
    fontsize=9
)

# ==========================================
# 8. FOOTNOTE
# ==========================================

fig.text(
    0.10,
    0.015,
    "Note: Dotted colored lines represent original ATE estimates. "
    "The dashed horizontal line indicates zero effect. "
    "Values are simulated sensitivity estimates, not observed effects.",
    fontsize=8,
    color="#555555"
)

plt.tight_layout(rect=[0, 0.06, 1, 1])

plt.savefig(
    "sensitivity.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

```
